# SentiSense — Exploratory Data Analysis

Quick look at the IMDb 50K dataset before we train.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sentisense.data import load_dataset
from sentisense.data.preprocessing import clean_text
```

```python
df = load_dataset()
df.head()
```

```python
df['label'].value_counts().plot(kind='bar', title='Class balance')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.show()
```

```python
df['review_length'] = df['text'].str.len()
df['review_length'].describe()
```

```python
sns.histplot(data=df, x='review_length', hue='label', bins=50, log_scale=True)
plt.title('Review length distribution by class (log scale)')
plt.show()
```

```python
# Inspect what clean_text does
for raw in df['text'].head(3):
    print('--- RAW ---')
    print(raw[:300])
    print('--- CLEANED ---')
    print(clean_text(raw)[:300])
    print()
```
